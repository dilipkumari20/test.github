```python
#Module 2 — Analytics Pipeline (/analytics)
#Part A — Profiling, cleaning, and the data story

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

#load the titanic dataset from seaborn
titanic = sns.load_dataset("titanic")

#save the loaded DataFrame as a committed offline fallback
titanic.to_csv("titanic.csv", index=False)
print("Titanic dataset saved successfully.")
#print the top 5 rows for checking data
print(titanic.head())
#print info() of dataset
print("\n--- DATA INFORMATION ---")
print(titanic.info())
#print statistical summary of dataset
print("\n--- STATISTICAL SUMMARY ---")
print(titanic.describe())
#print size of the dataset
print("\n--- DATASET SHAPE ---")
print(titanic.shape)
#Find missing values
print("\n--- MISSING VALUES ---")
missing_count = titanic.isnull().sum()
print(missing_count)
#missing percentage
missing_percentage = (titanic.isnull().sum() / len(titanic)) * 100
print("\nMissing value percentages:")
print(missing_percentage)
missing_percentage = missing_percentage[missing_percentage > 0]
print("\nColumns containing missing values:")
print(missing_percentage)
#Decide how to handle missing values
print("\n--- MISSING VALUE STRATEGY ---")

for column in missing_percentage.index:

    percentage = missing_percentage[column]

    print(f"{column}: {percentage:.2f}% missing")

    if percentage < 5:
        print("Strategy: Drop rows")

    elif percentage <= 30:
        print("Strategy: Impute")

    else:
        print("Strategy: High missingness - special decision required")

# Fill missing age values with the median age
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
# Treat missing deck values as their own category
titanic["deck"] = titanic["deck"].astype("object").fillna("Missing")
#Handle remaining missing rows
titanic = titanic.dropna()
print("\n--- MISSING VALUES AFTER CLEANING ---")

print(titanic.isnull().sum())

# Save cleaned data
titanic.to_csv("titanic.csv", index=False)

print("\nCleaned Titanic dataset saved.")

#Histogram of age
plt.figure(figsize=(8, 5))
plt.hist(titanic["age"], bins=30)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Distribution of Age")
plt.show()

#Histogram of fare
plt.figure(figsize=(8, 5))
plt.hist(titanic["fare"], bins=30)
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.title("Distribution of Fare")
plt.show()

#Box plot of age
plt.figure(figsize=(8, 5))
sns.boxplot(x=titanic["age"])
plt.title("Age Boxplot")
plt.show()

#Box plot of fare
plt.figure(figsize=(8, 5))
sns.boxplot(x=titanic["fare"])
plt.title("Fare Boxplot")
plt.show()

#IQR outliers for Age
Q1_age = titanic["age"].quantile(0.25)
Q3_age = titanic["age"].quantile(0.75)

IQR_age = Q3_age - Q1_age

lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age

age_outliers = titanic[
    (titanic["age"] < lower_age) |
    (titanic["age"] > upper_age)
]

print("Age outliers:", len(age_outliers))

#IQR outliers for Fare
Q1_fare = titanic["fare"].quantile(0.25)
Q3_fare = titanic["fare"].quantile(0.75)

IQR_fare = Q3_fare - Q1_fare

lower_fare = Q1_fare - 1.5 * IQR_fare
upper_fare = Q3_fare + 1.5 * IQR_fare

fare_outliers = titanic[
    (titanic["fare"] < lower_fare) |
    (titanic["fare"] > upper_fare)
]

print("Fare outliers:", len(fare_outliers))

#Fare mean, median and mode
fare_mean = titanic["fare"].mean()

fare_median = titanic["fare"].median()

fare_mode = titanic["fare"].mode()[0]

print("Fare Mean:", fare_mean)
print("Fare Median:", fare_median)
print("Fare Mode:", fare_mode)

# Fare Skewness
print("Fare skewness:", titanic["fare"].skew())
print("Mean > Median > Mode so, This indicates a right-skewed distribution.")

#Survival rate by sex
survival_by_sex = titanic.groupby("sex")["survived"].mean()
print("Survival rate by sex:")
print(survival_by_sex)
#Convert to percentage:
print("\nPercentage:")
print(survival_by_sex * 100)

#Survival rate by passenger class
survival_by_class = titanic.groupby("pclass")["survived"].mean()
print("Survival rate by passenger class:")
print(survival_by_class)
#convert into percentage
print("\nPercentage:")
print(survival_by_class * 100)

#Survival by sex AND pclass
survival_sex_class = titanic.groupby(
    ["sex", "pclass"]
)["survived"].mean()
print(survival_sex_class)
#Boolean masking
female_first = titanic[
    (titanic["sex"] == "female") &
    (titanic["pclass"] == 1)
]
print("Female first-class passengers:")
print(female_first["survived"].mean())

#Required correlation matrix
corr_columns = [
    "survived",
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare"
]

corr_matrix = titanic[corr_columns].corr()
print(corr_matrix)

#Correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Titanic Correlation Matrix")
plt.show()

#Find two strongest correlations
# Take absolute values
abs_corr = corr_matrix.abs()

# Remove diagonal
np.fill_diagonal(abs_corr.values, np.nan)

# Stack all pairs
strongest = abs_corr.stack().sort_values(ascending=False)

print("Strongest correlations:")

print(strongest.head(4))

pairs = []

for i in range(len(corr_columns)):
    for j in range(i + 1, len(corr_columns)):

        col1 = corr_columns[i]
        col2 = corr_columns[j]

        value = corr_matrix.loc[col1, col2]

        pairs.append((col1, col2, value, abs(value)))

pairs = sorted(
    pairs,
    key=lambda x: x[3],
    reverse=True
)

print("Top two strongest correlations:")

for pair in pairs[:2]:
    print(pair)

# Multivariate Chart 1 — Survival by Sex
plt.figure(figsize=(7, 5))

sns.barplot(
    data=titanic,
    x="sex",
    y="survived"
)

plt.title("Survival Rate by Sex")
plt.ylabel("Survival Rate")

plt.show()
'''Interpretation
The chart shows that female passengers had a much higher survival rate than male passengers.
A larger proportion of females survived the Titanic disaster, while the survival rate for males was considerably lower.
This indicates a clear difference in survival outcomes between the two sexes.'''

#Multivariate Chart 2 — Survival by Class
plt.figure(figsize=(7, 5))

sns.barplot(
    data=titanic,
    x="pclass",
    y="survived"
)

plt.title("Survival Rate by Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
''' Interpretation
The chart shows that passengers in 1st class had the highest survival rate, while 3rd-class passengers had the lowest survival rate.
Passengers in 2nd class had a survival rate between those of 1st and 3rd class.
Overall, survival rate decreased as the passenger class number increased.'''

# Multivariate Chart 3 — Sex + Class
plt.figure(figsize=(8, 5))

sns.barplot(
    data=titanic,
    x="pclass",
    y="survived",
    hue="sex"
)

plt.title("Survival by Passenger Class and Sex")
plt.ylabel("Survival Rate")

plt.show()
''' Interpretations
The chart shows that females had a higher survival rate than males across all passenger classes.
Among females, 1st-class passengers had the highest survival rate, while 3rd-class females had a lower survival rate.
For males, survival rates were generally lower, with 3rd-class males having the lowest survival rate.'''

#Multivariate Chart 4 — Age vs Fare
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    hue="survived"
)

plt.title("Age vs Fare by Survival")

plt.show()
''' Interpretation
The scatter plot shows the relationship between age, fare, and survival.
 Passengers who paid higher fares appear to have a higher proportion of survivors, while many passengers with lower fares did not survive. There is no clear direct relationship between age and survival, as both survivors and non-survivors are spread across different age groups.'''

#Standardize Age and Fare
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

titanic[["age_scaled", "fare_scaled"]] = scaler.fit_transform(
    titanic[["age", "fare"]]
)
print("Age scaled mean:")
print(titanic["age_scaled"].mean())

print("\nFare scaled mean:")
print(titanic["fare_scaled"].mean())

print("Age scaled std:")
print(titanic["age_scaled"].std())

print("\nFare scaled std:")
print(titanic["fare_scaled"].std())

print(titanic["age_scaled"].std(ddof=0))
print(titanic["fare_scaled"].std(ddof=0))



#PART B — MACHINE LEARNING
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, classification_report,
                             precision_score,recall_score,
                             confusion_matrix,
                             roc_auc_score,f1_score,roc_curve,
                             mean_absolute_error,
                             mean_squared_error,r2_score)
from imblearn.over_sampling import SMOTE
import joblib
# Train/Test split FIRST
X = titanic[
    [
        "pclass",
        "sex",
        "age",
        "sibsp",
        "parch",
        "fare",
        "embarked"
    ]
]

y = titanic["survived"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
print("Training class balance:")
print(y_train.value_counts(normalize=True))

print("\nTesting class balance:")
print(y_test.value_counts(normalize=True))
# Define numeric and categorical columns
numeric_features = [
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare"
]

categorical_features = [
    "sex",
    "embarked"
]

#Numeric preprocessing
numeric_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)
#Categorical preprocessing
categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

#ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_pipeline,
            numeric_features
        ),
        (
            "cat",
            categorical_pipeline,
            categorical_features
        )
    ]
)
#Logistic Regression Pipeline
logistic_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)
# Train the model
logistic_pipeline.fit(X_train, y_train)
# Prediction of the model
logistic_pred = logistic_pipeline.predict(X_test)

logistic_prob = logistic_pipeline.predict_proba(
    X_test
)[:, 1]

# Decision Tree Pipeline
tree_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            DecisionTreeClassifier(
                max_depth=5,
                random_state=42
            )
        )
    ]
)
# train the model
tree_pipeline.fit(X_train, y_train)
# prediction the model
tree_pred = tree_pipeline.predict(X_test)

tree_prob = tree_pipeline.predict_proba(
    X_test
)[:, 1]

# Random Forest Pipeline
forest_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestClassifier(
                n_estimators=200,
                random_state=42,
                oob_score=True
            )
        )
    ]
)
# Train the model
forest_pipeline.fit(X_train, y_train)
# Prediction the model
forest_pred = forest_pipeline.predict(X_test)

forest_prob = forest_pipeline.predict_proba(
    X_test
)[:, 1]

#Evaluation function
def evaluate_model(name, y_test, predictions, probabilities):

    accuracy = accuracy_score(y_test, predictions)

    precision = precision_score(
        y_test,
        predictions
    )

    recall = recall_score(
        y_test,
        predictions
    )

    f1 = f1_score(
        y_test,
        predictions
    )

    auc = roc_auc_score(
        y_test,
        probabilities
    )

    print("\n==============================")
    print(name)
    print("==============================")

    print("Accuracy :", accuracy)
    print("Precision:", precision)
    print("Recall   :", recall)
    print("F1 Score :", f1)
    print("AUC      :", auc)

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, predictions))

    return {
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "AUC": auc
    }

#Evaluate all models
results = []

results.append(
    evaluate_model(
        "Logistic Regression",
        y_test,
        logistic_pred,
        logistic_prob
    )
)

results.append(
    evaluate_model(
        "Decision Tree",
        y_test,
        tree_pred,
        tree_prob
    )
)

results.append(
    evaluate_model(
        "Random Forest",
        y_test,
        forest_pred,
        forest_prob
    )
)
#Comparison table
results_df = pd.DataFrame(results)

print(results_df)
#Confusion matrices
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

models_predictions = [
    ("Logistic Regression", logistic_pred),
    ("Decision Tree", tree_pred),
    ("Random Forest", forest_pred)
]

for ax, (name, predictions) in zip(
    axes,
    models_predictions
):

    cm = confusion_matrix(
        y_test,
        predictions
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.tight_layout()
plt.show()

# Roc curve
plt.figure(figsize=(8, 6))

fpr_log, tpr_log, _ = roc_curve(
    y_test,
    logistic_prob
)

fpr_tree, tpr_tree, _ = roc_curve(
    y_test,
    tree_prob
)

fpr_forest, tpr_forest, _ = roc_curve(
    y_test,
    forest_prob
)

plt.plot(
    fpr_log,
    tpr_log,
    label="Logistic Regression"
)

plt.plot(
    fpr_tree,
    tpr_tree,
    label="Decision Tree"
)

plt.plot(
    fpr_forest,
    tpr_forest,
    label="Random Forest"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curves")

plt.legend()

plt.show()

# Decision Tree visualization
X_train_transformed = tree_pipeline.named_steps[
    "preprocessor"
].transform(X_train)
#Get features
feature_names = (
    tree_pipeline
    .named_steps["preprocessor"]
    .get_feature_names_out()
)
# Plot decesion tree
plt.figure(figsize=(20, 10))

plot_tree(
    tree_pipeline.named_steps["model"],
    feature_names=feature_names,
    class_names=["Not Survived", "Survived"],
    filled=True,
    max_depth=3
)

plt.title("Decision Tree")

plt.show()

#Class balance
class_balance = y.value_counts()

print(class_balance)

print("\nClass balance percentage:")

print(
    y.value_counts(normalize=True) * 100
)
#Imbalance Strategy 1 — Baseline
baseline_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)

baseline_model.fit(X_train, y_train)

baseline_pred = baseline_model.predict(X_test)

#Evalution
baseline_precision = precision_score(
    y_test,
    baseline_pred
)

baseline_recall = recall_score(
    y_test,
    baseline_pred
)

baseline_f1 = f1_score(
    y_test,
    baseline_pred
)

print(
    "Baseline:",
    baseline_precision,
    baseline_recall,
    baseline_f1
)

#Imbalance Strategy 2 — class_weight balanced
balanced_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)

balanced_model.fit(
    X_train,
    y_train
)

balanced_pred = balanced_model.predict(
    X_test
)
#Evalution
balanced_precision = precision_score(
    y_test,
    balanced_pred
)

balanced_recall = recall_score(
    y_test,
    balanced_pred
)

balanced_f1 = f1_score(
    y_test,
    balanced_pred
)

print(
    "Balanced:",
    balanced_precision,
    balanced_recall,
    balanced_f1
)
#SMOTE
X_train_processed = preprocessor.fit_transform(
    X_train
)

X_test_processed = preprocessor.transform(
    X_test
)
smote = SMOTE(
    random_state=42
)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_processed,
    y_train
)
#Train the model
smote_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

smote_model.fit(
    X_train_smote,
    y_train_smote
)
#prediction of the model
smote_pred = smote_model.predict(
    X_test_processed
)
#Evalution
smote_precision = precision_score(
    y_test,
    smote_pred
)

smote_recall = recall_score(
    y_test,
    smote_pred
)

smote_f1 = f1_score(
    y_test,
    smote_pred
)

print(
    "SMOTE:",
    smote_precision,
    smote_recall,
    smote_f1
)
#Imbalance comparison table
imbalance_results = pd.DataFrame({
    "Strategy": [
        "Baseline",
        "Class Weight Balanced",
        "SMOTE"
    ],

    "Precision": [
        baseline_precision,
        balanced_precision,
        smote_precision
    ],

    "Recall": [
        baseline_recall,
        balanced_recall,
        smote_recall
    ],

    "F1": [
        baseline_f1,
        balanced_f1,
        smote_f1
    ]
})

print(imbalance_results)

#Random Forest GridSearchCV
rf = RandomForestClassifier(
    random_state=42,
    oob_score=True
)
rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", rf)
    ]
)
#parameter grid
param_grid = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [3, 5, 10, None],
    "model__max_features": ["sqrt", "log2"]
}
#Grid search
grid_search = GridSearchCV(
    rf_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

grid_search.fit(
    X_train,
    y_train
)
#Best parameter
print("Best parameters:")

print(grid_search.best_params_)
#Best CV score
print("Best CV score:")

print(grid_search.best_score_)
#OOB score
best_rf = grid_search.best_estimator_

oob_score = (
    best_rf
    .named_steps["model"]
    .oob_score_
)

print("OOB Score:", oob_score)

#Regression side task
regression_features = [
    "pclass",
    "sex",
    "age",
    "sibsp",
    "parch",
    "embarked"
]

X_reg = titanic[regression_features]

y_reg = titanic["fare"]

#split the data
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42
)
#Regression preprocessing
reg_numeric = [
    "pclass",
    "age",
    "sibsp",
    "parch"
]

reg_categorical = [
    "sex",
    "embarked"
]
reg_numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)
reg_categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)
reg_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            reg_numeric_pipeline,
            reg_numeric
        ),
        (
            "cat",
            reg_categorical_pipeline,
            reg_categorical
        )
    ]
)
#Linear regression
regression_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            reg_preprocessor
        ),
        (
            "model",
            LinearRegression()
        )
    ]
)
# Train the model
regression_pipeline.fit(
    X_reg_train,
    y_reg_train
)
#prediction of the model
fare_pred = regression_pipeline.predict(
    X_reg_test
)
#Regression metrics
mae = mean_absolute_error(
    y_reg_test,
    fare_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_reg_test,
        fare_pred
    )
)

r2 = r2_score(
    y_reg_test,
    fare_pred
)
#Adjusted R²:
n = len(y_reg_test)

p = X_reg_test.shape[1]

adjusted_r2 = 1 - (
    (1 - r2) *
    (n - 1) /
    (n - p - 1)
)
print("MAE:", mae)

print("RMSE:", rmse)

print("R²:", r2)

print("Adjusted R²:", adjusted_r2)

#Residual plot

residuals = y_reg_test - fare_pred

plt.figure(figsize=(8, 5))

plt.scatter(
    fare_pred,
    residuals
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("Predicted Fare")
plt.ylabel("Residual")

plt.title("Regression Residual Plot")

plt.show()
'''Interpretation

If the residuals are randomly scattered around zero with roughly constant spread, there is no strong visual evidence of heteroscedasticity.

If the spread becomes much wider or narrower as predicted fare increases, that suggests heteroscedasticity.

The assignment explicitly requires you to state this conclusion.'''

#Final model comparison table
classification_table = results_df[
    [
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "AUC"
    ]
]

print("CLASSIFICATION RESULTS")
print(classification_table)
regression_table = pd.DataFrame({
    "Regression Model": ["Linear Regression"],
    "MAE": [mae],
    "RMSE": [rmse],
    "R2": [r2],
    "Adjusted R2": [adjusted_r2]
})

print("\nREGRESSION RESULTS")
print(regression_table)

#Save the complete pipeline
full_pipeline = grid_search.best_estimator_

joblib.dump(
    full_pipeline,
    "titanic_best_pipeline.joblib"
)

print("Complete pipeline saved successfully.")

#Reload the saved pipeline
loaded_pipeline = joblib.load(
    "titanic_best_pipeline.joblib"
)

print("Pipeline loaded successfully.")

#Test the saved pipeline on RAW data
sample_raw = X_test.head(5)

print(sample_raw)
#predictions
sample_predictions = loaded_pipeline.predict(
    sample_raw
)

print("Predictions:")
print(sample_predictions)

#Save cleaned CSV
titanic.to_csv(
    "titanic_cleaned.csv",
    index=False
)
```
